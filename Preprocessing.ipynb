```python
import pandas as pd
import kagglehub
import seaborn as sns
### Knihovna na scaler
from sklearn.preprocessing import StandardScaler

# Download latest version
path = kagglehub.dataset_download("kundanbedmutha/exam-score-prediction-dataset")

# Zakladni vykresleni
df = pd.read_csv(path + "/Exam_Score_Prediction.csv")

# Cilova promenna je exam_score
# Meneni yes no na 0 1
df["internet_access"] = df["internet_access"].map({"yes": 1, "no": 0})
df["gender"] = df["gender"].map({"male": 1, "female": 0, "other": 2})

# Pivotovani promenych co nejsou ciselne
df = pd.get_dummies(df, columns=["exam_difficulty", "course", "facility_rating"], drop_first=True)

# Razeni na numericke a ordinalni hodnoty
target_name = "exam_score"
numerical_variables = ["age", "study_hours", "class_attendance", "sleep_hours"]
ordinal_variables = ["exam_difficulty", "course", "facility_rating", "internet_access", "gender"]

# Zacisteni
df = df.dropna()
df = df.drop(columns=["student_id"])

# Scaler
scaler = StandardScaler()
df[numerical_variables] = scaler.fit_transform(df[numerical_variables])
# Vypsani
df
```
